```python
score1=float(input("Enter first score: "))
score2=float(input("Enter second score: "))
score3=float(input("Enter third score: "))

average=(score1+score2+score3)/3

if (average>=90):
  grade='A'
elif(average>=75):
  grade='B'
elif(average>=60):
  grade='C'
else:
  grade='F'

print(f"Average score: {average:.2f}")
print("Grade: ", grade)
```

```python
list_num = input("Enter numbers separated by commas: ")

#splits list of strings by comma
numbers_str = list_num.split(",")
numbers = []

#converts list of strings into ints
for num in numbers_str:
    numbers.append(int(num.strip()))

total = 0
largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    total += num
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num

average = total / len(numbers)

print("Sum:", total)
print("Largest:", largest)
print("Smallest:", smallest)
print("Average:", round(average, 2))
```

```python
# a loop to take number inputs from user and have the user choose the arethmatic
# op they want to perform on the numbers.
while True:
  n1=int(input("enter the first number: "))
  n2=int(input("enter the second number: "))
  op=input("enter the operator (+, -, *, /, %, **),(type exit to stop): ")
  if op=="+":
    result=n1+n2
  elif op=="-":
    result=n1-n2
  elif op=="*":
    result=n1*n2
  elif op=="/":
    if n2 == 0:
            print("Error: Cannot divide by zero")
            continue
    result=n1 / n2
  elif op=="%":
    result=n1%n2
  elif op=="**":
    result=n1**n2
  elif op=="exit":
    break
  else:
    print("Invalid operator")

  print(f"{n1} {op} {n2} = {result}")
```
